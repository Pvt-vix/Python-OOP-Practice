```python
#Classed and object

class car():

    def start(self):
        print("Car is Started")

    def stop(self):
        print("Car is Stopped")

car1 = car()
car2 = car()

car1.start()
car2.start()

car1.stop()
car2.stop()

```

```python
class Student:
    def set_details(self,name,age,roll_no):
        self.name = name
        self.age = age
        self.roll_no = roll_no

    def see_details(self):
        print(f"student is {self.name}, and the age is {self.age}, and the roll number is {self.roll_no}")

class1 = Student() 
class1.set_details("John", 20, 101)

class2 = Student()
class2.set_details("Alice", 22, 102)

class1.see_details()
class2.see_details()    
```

```python
class car:

    def set_details(self, brand, name, year):
        self.brand = brand
        self.name = name
        self.year = year

    def show_details(self):
         print(f"Car Brand: {self.brand}")
         print(f"Car Name: {self.name}")
         print(f"Car Year: {self.year}")

car1 = car()
car1.set_details("Toyota", "Camry", 2020)

car2 = car()
car2.set_details("Honda", "Civic", 2019)

car1.show_details()
car2.show_details()
```

```python
#code Without Constructors 

class car:
    def set_details(self,brand,color):
        self.brand = brand
        self.color = color

car1 = car()
car1.set_details("Toyota", "Red")

car2 = car()
car2.set_details("Honda", "Blue")

print(f"Car 1: Brand - {car1.brand}, Color - {car1.color}")
print(f"Car 2: Brand - {car2.brand}, Color - {car2.color}")
```

Code with Constructor __init__ ()

```python
class Car():

    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

car1 = car("Toyota", "Red") #Creating an object of the car class and passing the brand and color as arguments to the constructor

car2 = car("Honda", "Blue") #creating another object of the car class and passing the brand and color as arguments to the constructor

print(f"Car 1: Brand - {car1.brand}, Color - {car1.color}") #printing the brand and color of car1 using the attributes set by the constructor

print(f"Car 2: Brand - {car2.brand}, Color - {car2.color}") #printing the brand and color of car2 using the attributes set by the constructor
```
